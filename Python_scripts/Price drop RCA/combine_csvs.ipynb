```python
import pandas as pd
import os

# Define your file names
file1 = 'ppvs_jan_feb_mar_apr_2025.csv'
file2 = 'ppvs_may_jun_jul_aug_2025.csv'
file3 = 'ppvs_sep_oct_nov_dec_2025.csv'
output = 'PPVS_Unbranded_2025.csv'

def combine_specific_files():
    # Check if files exist to avoid that FileNotFoundError
    missing_files = [f for f in [file1, file2, file3] if not os.path.exists(f)]
    
    if missing_files:
        print(f"❌ Error: Could not find these files: {missing_files}")
        print(f"Current working directory: {os.getcwd()}")
        print("💡 Tip: Make sure your CSVs are in the folder listed above.")
        return

    print(f"✅ Found {file1}, {file2} and {file3} Combining now...")

    # Load the files
    df1 = pd.read_csv(file1)
    df2 = pd.read_csv(file2)
    df3 = pd.read_csv(file3)

    # Combine them (stacking sales2 under sales1)
    combined_df = pd.concat([df1, df2, df3], ignore_index=True)

    # Save the result
    combined_df.to_csv(output, index=False)
    print(f"🚀 Success! Combined file saved as: {output}")

if __name__ == "__main__":
    combine_specific_files()
```
