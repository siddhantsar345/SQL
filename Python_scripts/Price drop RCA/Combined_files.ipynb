```python
import pandas as pd
import os

# All 12 monthly files in chronological order
files = [
    'PPVS_Branded_Jan_2025.csv',
    'PPVS_Branded_Feb_2025.csv',
    'PPVS_Branded_Mar_2025.csv',
    'PPVS_Branded_Apr_2025.csv',
    'PPVS_Branded_May_2025.csv',
    'PPVS_Branded_June_2025.csv',
    'PPVS_Branded_July_2025.csv',
    'PPVS_Branded_Aug_2025.csv',
    'PPVS_Branded_Sept_2025.csv',
    'PPVS_Branded_Oct_2025.csv',
    'PPVS_Branded_Nov_2025.csv',
    'PPVS_Branded_Dec_2025.csv',
]
output = 'PPVS_Branded_2025.csv'

def combine_specific_files():
    # Check if files exist to avoid that FileNotFoundError
    missing_files = [f for f in files if not os.path.exists(f)]

    if missing_files:
        print(f"❌ Error: Could not find these files: {missing_files}")
        print(f"Current working directory: {os.getcwd()}")
        print("💡 Tip: Make sure your CSVs are in the folder listed above.")
        return

    print(f"✅ Found all {len(files)} files. Combining now...")

    # Load and combine in one go
    combined_df = pd.concat(
        [pd.read_csv(f) for f in files],
        ignore_index=True
    )

    # Save the result
    combined_df.to_csv(output, index=False)
    print(f"🚀 Success! Combined file saved as: {output}")
    print(f"📊 Total rows: {len(combined_df):,}")

if __name__ == "__main__":
    combine_specific_files()
```
