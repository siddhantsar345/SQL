```python
import pandas as pd

# Load your CSV file
file_path = 'PPVS_Branded_Aug_2025.csv'  # Replace with your actual file name
df = pd.read_csv(file_path)

# 1. Identify the last column dynamically
last_col = df.columns[-1]

# 2. Convert the column to datetime objects
# Format '%Y %m %d' matches your "2026 04 01" structure
df[last_col] = pd.to_datetime(df[last_col], format='%Y %m %d')

# 3. Find the Start (min) and End (max) dates
start_date = df[last_col].min()
end_date = df[last_col].max()

# 4. Display the results
print(f"Analyzing column: {last_col}")
print(f"Start Date: {start_date.strftime('%Y-%m-%d')}")
print(f"End Date:   {end_date.strftime('%Y-%m-%d')}")
```
