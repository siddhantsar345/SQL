```python
import pandas as pd

# --- Configuration ---
file_path = 'Traffic_vertical_2025.csv'  # <-- Replace with your actual file name

# --- Load the CSV ---
try:
    df = pd.read_csv(file_path)
except FileNotFoundError:
    raise SystemExit(f"Could not find the file: {file_path}")

if df.empty:
    raise SystemExit("The CSV has no rows.")

# 1. Get the second column
date_col = df.columns[1]
print(f"Analyzing column: {date_col}")

# 2. Convert the column to datetime objects.
#    order_date_key is a YYYYMMDD "date key". Missing values can force the
#    column to float (e.g. 20260401.0), so convert to a nullable integer
#    first, then to a clean 8-char string, then parse. errors='coerce'
#    turns anything unparseable into NaT instead of crashing.
keys = pd.to_numeric(df[date_col], errors='coerce').astype('Int64').astype(str)
dates = pd.to_datetime(keys, format='%Y%m%d', errors='coerce')

# 3. Report any rows that could not be parsed
bad_rows = dates.isna().sum()
if bad_rows:
    print(f"Warning: {bad_rows} value(s) could not be parsed as a date and were ignored.")

valid = dates.dropna()
if valid.empty:
    raise SystemExit(f"No valid dates found in column '{date_col}'. Check the date format.")

# 4. Find the Start (min) and End (max) dates
start_date = valid.min()
end_date = valid.max()

# 5. Display the results
print(f"Start Date: {start_date.strftime('%Y-%m-%d')}")
print(f"End Date:   {end_date.strftime('%Y-%m-%d')}")
print(f"Span:       {(end_date - start_date).days} days")
```
